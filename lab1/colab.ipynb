```python
from importlib.metadata import PackageNotFoundError, version

for name in ("numpy", "pandas", "matplotlib", "seaborn", "scikit-learn"):
    try:
        print(f"{name:<15} {version(name)}")
    except PackageNotFoundError:
        print(f"{name:<15} отсутствует")
# При необходимости: !pip install seaborn
```

### TPU
Выберите TPU в том же меню. В новом сеансе запустите только ячейку TPU, чтобы сохранить вывод GPU.

```python
# TPU — выполнить после смены T4 на TPU v5e-1
import os
os.environ["JAX_PLATFORMS"] = "tpu"
import jax

print("JAX:", jax.__version__)
print("Устройств:", jax.device_count())
print("Процессов:", jax.process_count())
for device in jax.devices("tpu"):
    print(f"{device} | модель: {device.device_kind} | координаты: {device.coords} | ядро: {device.core_on_chip}")
```
